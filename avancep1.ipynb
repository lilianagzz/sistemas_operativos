```python
import threading
import subprocess
import shlex
import time

procesos_iniciados = []   # guardaremos aquí solo los PID que la app inicia

# ---------- Funciones básicas ----------
def listar():
    subprocess.run(["ps", "-eo", "pid,comm,pcpu,pmem", "--no-header"])

def iniciar():
    cmd = input("Comando a iniciar: ")
    try:
        args = shlex.split(cmd)
        p = subprocess.Popen(args)
        procesos_iniciados.append(p.pid)
        print(f"Proceso iniciado con PID {p.pid}")
    except FileNotFoundError:
        print("Comando no encontrado")

def detener():
    pid = input("PID a detener: ")
    if pid.isdigit() and int(pid) in procesos_iniciados:
        subprocess.run(["kill", pid])
        print("Proceso detenido")
    else:
        print("PID no válido o no fue iniciado por esta app")

# ---------- Monitoreo en hilo ----------
def monitoreo_hilo():
    for _ in range(5):
        print("\nCPU y Memoria:")
        subprocess.run(["ps", "-eo", "pid,pcpu,pmem", "--no-header"])
        time.sleep(1)

def monitorear():
    hilo = threading.Thread(target=monitoreo_hilo)
    hilo.start()
    hilo.join()

# ---------- Menú ----------
while True:
    print("\n1) Listar procesos")
    print("2) Iniciar proceso")
    print("3) Detener proceso")
    print("4) Monitorear CPU/Memoria (hilo)")
    print("0) Salir")
    op = input("Opción: ")

    if op == "1": listar()
    elif op == "2": iniciar()
    elif op == "3": detener()
    elif op == "4": monitorear()
    elif op == "0": break
    else: print("Opción no válida")
```
